# Text Generation with Google T5 and mT5
Reference: https://huggingface.co/google-t5/t5-base

```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
import torch
```

```python
tokenizer = AutoTokenizer.from_pretrained("./Model/Google-T5")
model = AutoModelForSeq2SeqLM.from_pretrained("./Model/Google-T5")

device = torch.device("cuda" if torch.cuda.is_available() else "mps")
model.to(device)
model.eval()
```

```python
input_ids = tokenizer("Translate English to German: The weather is nice today.", return_tensors="pt").to(model.device)
output = model.generate(**input_ids, cache_implementation="static")
print(tokenizer.decode(output[0], skip_special_tokens=True))
```
